```python
import pymysql

conn = pymysql.connect(
    database = 'mysite',
    user = 'root',
    password = 'mysql',
    host = 'localhost',
    port = 3306,
)

try:
    with conn.cursor() as cursor:
        # 데이터 삽입 SQL
        insert_query = 'insert into bookmark_bookmark(title, url) values (%s, %s)'

        # 삽입할 데이터 리스트
        bookmarks = [
            ("Google", "https://www.google.com"),
            ("Youtube", "https://www.youtube.com"),
            ("Github", "https://www.github.com"),
            ("Stack Overflow", "https://stackoverflow.com"),
            ("Django Docs", "https://docs.Djangoproject.com"),
            ("Python Docs", "https://docs.python.com"),
            ("CNN", "https://www.cnn.coim.com"),
            ("NAVER", "https://www.naver.com"),
            ("KAKAO", "https://www.kakao.com"),
            ("DAUM", "https://www.daum.net")

        ]

        # 여러개의 레코드를 삽입
        cursor.executemany(insert_query, bookmarks)

        # 변경사항 저장
        conn.commit()
        print("데이터 삽입 완료!")

except Exception as e:
    print(f"오류 발생 : {e}")
    conn.rollback()
finally:
    conn.close()
```

```python
import pandas as pd

conn = pymysql.connect(
    database = 'mysite',  # 생성한 MySQL 데이터베이스 이름
    user = 'root',  # MySQL 사용자명
    password = 'mysql',  # MySQL 비밀번호
    host = 'localhost',
    port = 3306
)

try:
    with conn.cursor() as cursor:
        #select 쿼리 실행
        cursor.execute("SELECT * FROM bookmark_bookmark")
        results = cursor.fetchall()

        #결과를 DataFrame으로 변환
        df = pd.DataFrame(results, columns=['id', 'title', 'url'])
        display(df)
finally:
    conn.close()
```
