```python
# 0. 라이브러리 임포트

import pandas as pd
import oracledb
```

```python
# 1. 데이터 프레임 생성

df = pd.DataFrame({
    'id': [1, 2, 3],
    'name': ['Alice', 'Bob', 'Charlie'],
    'age': [25, 30, 35]
})

print(df)
```

```python
# 2. Oracle연결 (DB Connection 생성)

conn = oracledb.connect(
    user='ora_user',
    password='1234',
    dsn='localhost/XEPDB1'          # 호스트명/서비스명
)

cursor = conn.cursor()
print("DB연결 성공!")
```

```python
# 3. INSERT문 준비

insert_sql = "INSERT INTO test_user(id, name, age) VALUES(:1, :2, :3)"
                                # 1번 변수, 2번 변수, 3번 변수가 순서대로 맞춰서 들어감 / Oracle 쿼리 형식

data = [tuple(x) for x in df.to_numpy()]

# print(data)
# [(1, 'Alice', 25), (2, 'Bob', 30), (3, 'Charlie', 35)]
```

```python
# 4. 데이터 삽입

cursor.executemany(insert_sql, data)
```

```python
# 5. 커밋 및 연결 종료

conn.commit()
cursor.close()
conn.close()
```
